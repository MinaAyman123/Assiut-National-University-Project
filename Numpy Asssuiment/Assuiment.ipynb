### 🧪 Assignment 1: Build Your Own `train_test_split_np()` Using NumPy


```python
import numpy as np
```

```python

def train_test_split_np(X, y, test_size=0.2, seed=None, shuffle=True):
    """
    Split arrays or matrices into random train and test subsets using NumPy only.

    Parameters
    ----------
    X : array-like, features
    y : array-like, labels
    test_size : float, default=0.2 → % of data for testing
    seed : int, random seed for reproducibility
    shuffle : bool, whether to shuffle data before splitting

    Returns
    -------
    X_train, X_test, y_train, y_test
    """

    if not (0 < test_size < 1):
        raise ValueError("test_size must be between 0 and 1")

    n_samples = X.shape[0]
    n_test = int(n_samples * test_size)
    n_train = n_samples - n_test

    if seed is not None:
        np.random.seed(seed)

    indices = np.arange(n_samples)
    if shuffle:
        np.random.shuffle(indices)

    train_idx = indices[:n_train]
    test_idx = indices[n_train:]

    return X[train_idx], X[test_idx], y[train_idx], y[test_idx]

```

### 📈 Assignment 2: Build Linear Regression from Scratch (Using Your Train-Test split)

```python
# Step 1: Generate dataset

np.random.seed(30)
X = 2 * np.random.rand(100, 1)          # Features
y = 4 + 3 * X + np.random.randn(100, 1) # Labels with noise

# Add bias term (x0 = 1)
X = np.insert(X, 0, 1, axis=1)

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Step 2: Split data into train & test
X_train, X_test, y_train, y_test = train_test_split_np(X, y, test_size=0.2, seed=42, shuffle=True)

print("Train size:", X_train.shape[0])
print("Test size:", X_test.shape[0])
```

```python
# Step 3: Train model using Normal Equation
Theta = np.linalg.inv(X_train.T @ X_train) @ (X_train.T @ y_train)

print("Theta (weights):\n", Theta)
```
